```python
### 1.2.1 Original Prompt

# e.g Generate a Python script to plot a 2D Gaussian function 
# using Numpy and Matplotlib
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Create a 2D grid
x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)
X, Y = np.meshgrid(x, y)

# Gaussian parameters
mu_x = 0
mu_y = 0
sigma_x = 1
sigma_y = 1

# 2D Gaussian function
Z = np.exp(
    -(
        ((X - mu_x) ** 2) / (2 * sigma_x**2)
        + ((Y - mu_y) ** 2) / (2 * sigma_y**2)
    )
)

# Plot the Gaussian
plt.figure(figsize=(8, 6))
plt.imshow(
    Z,
    extent=[-5, 5, -5, 5],
    origin="lower",
    cmap="viridis"
)

plt.colorbar(label="Gaussian value")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("2D Gaussian Function")
plt.show()
```

```python
import torch
import numpy as np

# Device configuration
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Use NumPy to create a 2D array of complex numbers on [-2,2]x[-2,2]
Y, X = np.mgrid[-1.3:1.3:0.005, -2:1:0.005]

# load into PyTorch tensors
x = torch.Tensor(X)
y = torch.Tensor(Y)
z = torch.complex(x, y)  # important!
zs = z.clone()
ns = torch.zeros_like(z)

# transfer to the GPU device
z = z.to(device)
zs = zs.to(device)
ns = ns.to(device)

# Mandelbrot Set
for i in range(200):
    zs_ = zs * zs + z
    not_diverged = torch.abs(zs_) < 4.0
    ns += not_diverged
    zs = zs_

# plot
import matplotlib.pyplot as plt
fig = plt.figure(figsize=(16, 10))

def processFractal(a):
    """Display an array of iteration counts as a colorful picture of a fractal."""
    a_cyclic = (6.28 * a / 20.0).reshape(list(a.shape) + [1])
    img = np.concatenate([10 + 20 * np.cos(a_cyclic),
                           30 + 50 * np.sin(a_cyclic),
                           155 - 80 * np.cos(a_cyclic)], 2)
    img[a == a.max()] = 0
    a = img
    a = np.uint8(np.clip(a, 0, 255))
    return a

plt.imshow(processFractal(ns.cpu().numpy()))
plt.tight_layout(pad=0)
plt.show()
```

```python
import torch
import matplotlib.pyplot as plt

# -----------------------------
# Configuration
# -----------------------------
WIDTH = 1600
HEIGHT = 1200
MAX_ITER = 500

# Mandelbrot region
X_MIN, X_MAX = -2.5, 1.0
Y_MIN, Y_MAX = -1.25, 1.25

# Use GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# -----------------------------
# Create complex coordinate grid
# -----------------------------
x = torch.linspace(X_MIN, X_MAX, WIDTH, device=device)
y = torch.linspace(Y_MIN, Y_MAX, HEIGHT, device=device)

yy, xx = torch.meshgrid(y, x, indexing="ij")
c = torch.complex(xx, yy)

# z starts at 0 for every point
z = torch.zeros_like(c)

# Stores the iteration at which each point escapes
iterations = torch.zeros(c.shape, dtype=torch.int32, device=device)

# Points that are still potentially in the Mandelbrot set
active = torch.ones(c.shape, dtype=torch.bool, device=device)

# -----------------------------
# Mandelbrot iteration
# -----------------------------
for i in range(MAX_ITER):
    # z = z^2 + c
    z[active] = z[active] ** 2 + c[active]

    # Points outside radius 2 have escaped
    escaped = torch.abs(z) > 2

    # Record the first iteration where a point escaped
    newly_escaped = escaped & active
    iterations[newly_escaped] = i

    # Only continue iterating points that haven't escaped
    active &= ~escaped

    # Stop early if everything escaped
    if not active.any():
        break

# Points that never escaped are inside the Mandelbrot set
iterations[active] = MAX_ITER

# Move result back to CPU for plotting
image = iterations.cpu().numpy()

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(12, 9))
plt.imshow(
    image,
    extent=[X_MIN, X_MAX, Y_MIN, Y_MAX],
    origin="lower",
    cmap="magma"
)

plt.xlabel("Re(c)")
plt.ylabel("Im(c)")
plt.title(f"Mandelbrot Set — {WIDTH} × {HEIGHT}, {MAX_ITER} iterations")
plt.colorbar(label="Escape iteration")
plt.show()
```
